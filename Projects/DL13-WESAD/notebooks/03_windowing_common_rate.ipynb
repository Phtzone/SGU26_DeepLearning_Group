# 03 — Chia cửa sổ Common-rate 14 kênh

Đọc kết quả notebook 02, tạo cửa sổ **30 giây / stride 15 giây / 64 Hz**.
Chỉ giữ Baseline, Stress và Amusement; cửa sổ phải nằm trọn một đoạn nhãn và không chạm `quality_mask=False`.
Đầu ra `[N, 14, 1920]`, chưa chuẩn hóa, chưa huấn luyện. Không thay đổi dữ liệu gốc hay thí nghiệm cũ.

Lưới stride bắt đầu lại ở đầu mỗi vùng nhãn hợp lệ; không đổi lưới khi loại cửa sổ do mask.
Chạy lại sẽ kiểm tra cache và tạo log/báo cáo mới. Xem `docs/Common_rate_windowing.md` để biết cấu trúc file.


```python
from pathlib import Path
import json
import subprocess
import sys
from datetime import datetime, timezone
from uuid import uuid4

root = next(p for p in [Path.cwd(), *Path.cwd().parents]
            if (p / 'src/data/common_rate_windowing.py').exists())
run_id = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ') + '_' + uuid4().hex[:8]
report = root / 'results/common_rate_windowing' / run_id
subprocess.run([sys.executable, '-m', 'scripts.prepare_common_rate_windows',
                '--report', str(report)], cwd=root, check=True)
print('Báo cáo:', report)
```

```python
import pandas as pd

summary = json.loads((report / 'summary.json').read_text(encoding='utf-8'))
verification = json.loads((report / 'verification.json').read_text(encoding='utf-8'))
display(pd.read_csv(report / 'windows_by_subject.csv'))
print('Tổng:', summary['totals'])
print('Đối chiếu tất cả cửa sổ đạt:', verification['all_passed'])
print('Log:', report / 'windowing.log')
print('Cửa sổ bị loại:', report / 'rejected_quality.csv')
print('Dữ liệu:', summary['config']['output'])
```
